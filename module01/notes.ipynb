# random arrays #

```python
np.random.rand(5, 2) # generate a random multi-dimensional array with 5 rows and 2 columns and elements betweem 0 and 1 
```

```python
np.random.seed(110) # note: the seed ensures you get the same sequence of random nums each time you re-run this cell/script from the top.
# Think of it as a bookmark that opens to Page 110 and puts your finger at Line 1.
# Calling rand() moves your finger down the page, so consecutive calls read the NEXT numbers—not the same ones—unless you reset the seed.
np.random.rand(5) # reads Lines 1–5 (not printed, moves finger to Line 6)
print(np.random.rand(5)) # reads Lines 6–10 (moves finger to Line 11)
print(np.random.rand(5)) # reads Lines 11–15 (different from above because your finger kept moving down the page)
```

```python
# The seed is like a bookmark: it opens the random-number book to a specific page (Page 110)
# and puts your finger at Line 1.
# "Same each time" means every time you re-run this entire cell/script from the top,
# the whole sequence below will play out identically.
np.random.seed(110)

# Reads Lines 1–5 on Page 110 (not printed, so these 5 numbers are used up)
np.random.rand(5)

# Continues reading from where your finger left off: Lines 6–10
print("Lines 6–10: ", np.random.rand(5))

# Continues reading forward: Lines 11–15 (different from above, because you didn't reset the bookmark)
print("Lines 11–15:", np.random.rand(5))

# To get the exact same array again, reset the seed to flip back to Line 1 of Page 110:
np.random.seed(110)
print("Back to Lines 1–5:", np.random.rand(5))
```

##### vector-vector multiplication #####

```python
# write a normal python functiuon that takes 2 vectors u and v and multiplies
# them together and returns the dot product. note without using numpy

u_vec = [1, 2, 3, 4]
v_vec = [5, 6, 7, 8]

def vect_vect_multiplication(u, v):
    assert len(u) == len(v), "Vectors must be the same length"
    res = []
    for i in range(len(u)):
        res.append(u[i] * v[i])
    
    dot_prod = sum(res)
    return dot_prod

print(vect_vect_multiplication(u_vec, v_vec)) 
```

when multiplying two vectors together eg u=[1,2,3,5] and v = [5,6,7,8] (dot product) we multiply each element in u with its corresponding element in v and then sum the anser, so 1*5, 2*6, 3*7, 5*8 = 5+12+21+40 = 78 (rember that vectors in liear algebra are usually written as columns)

We sum the products because the goal of a dot product is to collapse two lists of numbers into a single summary score that measures how much those two vectors agree or align with each other.

Imagine you are at a grocery store:
• Vector u (Quantities): [1 apple, 2 bananas, 3 oranges, 5 pears]
• Vector v (Prices): [£5, £6, £7, £8]
If you only multiply the matching elements ([5, 12, 21, 40]), you just have the subtotal for each fruit. Summing them (78) gives you the single total bill at the checkout.


Geometric Meaning: Measuring "Alignment"

In geometry, data science, and machine learning, vectors represent direction and strength (magnitude). Summing the multiplied pairs boils that comparison down to one number:
• Large positive sum: The two vectors point in the same general direction (strong agreement). For example, if a user likes sci-fi and action ([5, 4]) and a movie is heavy on sci-fi and action ([5, 5]), the dot product is high (45), so Netflix recommends it.
• Zero sum: The vectors are completely perpendicular (90 degrees apart)—they have zero overlap or relationship.
• Negative sum: The vectors point in opposite directions (disagreement).



Quick Contrast in NumPy

• Element-wise multiplication (u * v): Keeps the separate parts (array([5, 12, 21, 40])).
• Dot product (np.dot(u, v) or u @ v): Sums those parts into one total (78).


A zero sum means the two vectors are completely unrelated or perpendicular (pointing at a 90° angle), while a negative sum means they point in opposite directions (direct opposites).
To make this concrete, imagine a movie recommendation system where preferences are rated from -5 (hate) to +5 (love) across two genres: [Sci-Fi, Romance].

1. Zero Sum (No Relationship / 90° Apart)

Imagine User A loves Sci-Fi (5) and feels neutral about Romance (0). Now imagine Movie X has zero Sci-Fi (0) and is pure Romance (5).
• User A: [5, 0]
• Movie X: [0, 5]
• Dot Product: (5 * 0) + (0 * 5) = 0 + 0 = 0
What it means: There is zero overlap between what the user wants and what the movie offers. Visually, if you draw [5, 0] pointing straight East and [0, 5] pointing straight North, they form a 90-degree right angle—one does not move in the direction of the other at all.

2. Negative Sum (Opposite Directions)

Now imagine User B hates Sci-Fi (-5) and hates Romance (-4), while Movie Y is packed with Sci-Fi (5) and Romance (4).
• User B: [-5, -4]
• Movie Y: [5, 4]
• Dot Product: (-5 * 5) + (-4 * 4) = -25 + (-16) = -41
What it means: Everything the movie offers is the exact opposite of what the user wants. Visually, one arrow points North-East and the other points South-West—they are pulling in opposite directions.

Quick Summary Rule

• Positive (> 0): Pointing in a similar direction (good match).
• Zero (= 0): Perpendicular / 90° apart (completely independent, no overlap).
• Negative (< 0): Pointing in opposite directions (mismatch / clash).
